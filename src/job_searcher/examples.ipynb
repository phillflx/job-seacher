# LinkedIn job search example

This notebook authenticates through `SessionManager`, searches LinkedIn for a title, and displays the returned jobs. Run it from an environment where this project and its dependencies are installed.

```python
from dataclasses import asdict
from getpass import getpass

from job_searcher.job_search import JobSearch
from job_searcher.session_manager import SessionManager
```

## Sign in

The credentials are requested interactively and are not stored in the notebook. `SessionManager` opens Chrome; complete any LinkedIn verification prompts in that window.

```python
session = SessionManager()
session.authenticate(
    user=input("LinkedIn email: "),
    password=getpass("LinkedIn password: "),
)
job_search = JobSearch(session)
```

## Search jobs

Set a title and result limit, then run the search.

```python
job_title = "Data Engineer"
limit = 10
jobs = job_search.search(job_title, limit)
len(jobs)
```

```python
jobs
```

```python
# Display all collected fields for each listing.
[asdict(job) for job in jobs]
```
